```python
import os
os.makedirs("../data/text_files",exist_ok=True)
```

```python
sample_text = {
    "../data/text_files/python_intro.txt" : "Python is a high-level, interpreted, and general-purpose programming language known for its simplicity and readability. It was created by Guido van Rossum and first released in 1991. Python supports multiple programming paradigms, including procedural, object-oriented, and functional programming. Due to its extensive standard library and large ecosystem of third-party packages, Python is widely used in web development, data science, artificial intelligence, machine learning, automation, cybersecurity, and software development. Its easy-to-understand syntax makes it one of the most popular programming languages for both beginners and professional developers.",
    "../data/text_files/machine_learning.txt" : "Machine Learning (ML) is a branch of Artificial Intelligence (AI) that enables computers to learn from data and improve their performance without being explicitly programmed. Instead of following fixed rules, machine learning algorithms identify patterns in data and use those patterns to make predictions or decisions. Machine learning is widely used in applications such as recommendation systems, image recognition, speech processing, fraud detection, healthcare, and autonomous vehicles. By analyzing large amounts of data, machine learning helps organizations automate tasks, gain insights, and make more accurate decisions."
}

for file_path,file_content in sample_text.items():
    with open(file_path,'w', encoding="utf-8") as f:
        f.write(file_content)

print("file created successfully")
```

```python
### text loader to load and read the text file

from langchain_community.document_loaders import TextLoader
loader = TextLoader("../data/text_files/python_intro.txt",encoding="utf-8")
document = loader.load()
print(document)
```

```python
###  Directory loader means load or read all the text files in directory.
from langchain_community.document_loaders import DirectoryLoader

directory_load = DirectoryLoader(
    "../data/text_files",
    glob="**/*.txt", ## pattern to match all the files
    loader_cls=TextLoader,
    loader_kwargs={'encoding' : 'utf-8'},
    show_progress=False
)
directory = directory_load.load()
print(directory)
```
